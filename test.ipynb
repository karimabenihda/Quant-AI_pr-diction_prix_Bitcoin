```python
from pyspark.sql import SparkSession
import os


jar_path = os.path.abspath("postgresql-42.7.3.jar")

spark = SparkSession.builder \
    .appName("GoldToPostgres") \
    .config("spark.jars", jar_path) \
    .getOrCreate()


gold_df = spark.read.parquet("Machine_Learning/data/gold_dataset")

gold_df.show(10)
gold_df.printSchema()
```

```python
gold_df.write \
    .format("jdbc") \
    .option("url", "jdbc:postgresql://localhost:5432/bitcoin_db") \
    .option("dbtable", "bitcoin_gold") \
    .option("user", "app_user") \
    .option("password", "strongpassword") \
    .option("driver", "org.postgresql.Driver") \
    .mode("append") \
    .save()
```
